## Interpretation guardrails
- HaGRID trains gesture control; KIMORE trains the experimental quality estimator integrated into the app.
- UI-PRMD evaluates deterministic/personalized DTW scoring; it does not train the live pose estimator.
- KIMORE evaluation is subject-disjoint; live MediaPipe inference is cross-sensor and remains experimental.
- Runtime bars cover isolated CPU algorithms, not full camera-to-screen latency.
- Dataset accuracy is not clinical efficacy.
